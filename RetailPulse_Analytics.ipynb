### Customer segmentation (RFM analysis)
RFM stands for:
- Recency: How recently a customer purchased.
- Frequency: How many orders a customer placed.
- Monetary: How much money a customer spent.

### Calculate RFM metrics

```python
import pandas as pd
import numpy as np

# Reference date: one day after the last transaction
reference_date = sales_clean["Order_Date"].max() + pd.Timedelta(days=1)

rfm = sales_clean.groupby("Customer_ID").agg(
    Recency=("Order_Date", lambda x: (reference_date - x.max()).days),
    Frequency=("Order_ID", "nunique"),
    Monetary=("Sales", "sum")
).reset_index()

display(rfm.head(10))
```

### Prepare the data for clustering

```python
from sklearn.preprocessing import StandardScaler

# Reduce skew in customer spending
X = rfm[["Recency", "Frequency", "Monetary"]].copy()
X["Monetary"] = np.log1p(X["Monetary"])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Scaled data shape:", X_scaled.shape)
```

### Apply K-Means clustering

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

rfm["Segment"] = kmeans.fit_predict(X_scaled)

print("Customer segments created successfully!")
display(rfm.head(10))
```

### Interpret the customer segments

```python
segment_summary = rfm.groupby("Segment").agg(
    Customers=("Customer_ID", "count"),
    Avg_Recency=("Recency", "mean"),
    Avg_Frequency=("Frequency", "mean"),
    Avg_Spending=("Monetary", "mean")
).round(2)

display(segment_summary)
```

### Visualize the segments

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 6))

for segment in sorted(rfm["Segment"].unique()):
    subset = rfm[rfm["Segment"] == segment]
    plt.scatter(
        subset["Frequency"],
        subset["Monetary"],
        label=f"Segment {segment}",
        alpha=0.7
    )

plt.title("RetailPulse - Customer Segmentation")
plt.xlabel("Order Frequency")
plt.ylabel("Total Spending (₹)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Demand forecasting
Goal is to use historical sales quantities to estimate demand for the next 30 days and evaluate how well our baseline forecast performs

### Prepare daily demand data

```python
daily_demand = (
    sales_clean
    .groupby("Order_Date")["Quantity"]
    .sum()
    .resample("D")
    .sum()
    .asfreq("D", fill_value=0)
)

print("Start date:", daily_demand.index.min())
print("End date:", daily_demand.index.max())
display(daily_demand.head())
```

### Split the data into training and testing sets
I will reserve the final 30 days for testing, so the model is evaluated on dates it hasn't trained on.

```python
train = daily_demand.iloc[:-30]
test = daily_demand.iloc[-30:]

print("Training days:", len(train))
print("Testing days:", len(test))
```

### Create a baseline forecast
This baseline predicts demand using the average daily quantity from the last seven training days.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

forecast_value = train.tail(7).mean()

predictions = pd.Series(
    forecast_value,
    index=test.index
)

print("Predicted average daily demand:", round(forecast_value, 2))
```

### Evaluate the forecast
I will calculate Mean Absolute Error (MAE) and Mean Absolute Percentage Error (MAPE). 
MAPE is calculated only on test days with nonzero actual demand.

MAE measures the average absolute prediction error in units. 
MAPE expresses error as a percentage, but it is undefined for actual values of zero.

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(test, predictions)

nonzero_days = test > 0

if nonzero_days.any():
    mape = (
        np.abs(
            (test[nonzero_days] - predictions[nonzero_days])
            / test[nonzero_days]
        ).mean() * 100
    )
    print(f"MAE: {mae:.2f} units")
    print(f"MAPE on nonzero-demand days: {mape:.2f}%")
else:
    print(f"MAE: {mae:.2f} units")
    print("MAPE cannot be calculated because all test days have zero demand.")
```

### Plot actual versus predicted demand

```python
plt.figure(figsize=(12, 5))

plt.plot(test.index, test.values, label="Actual Demand", marker="o")
plt.plot(predictions.index, predictions.values,
         label="Baseline Forecast", linestyle="--")

plt.title("RetailPulse - Demand Forecast Evaluation")
plt.xlabel("Date")
plt.ylabel("Units Sold")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Forecast the next 30 days

```python
future_dates = pd.date_range(
    start=daily_demand.index.max() + pd.Timedelta(days=1),
    periods=30,
    freq="D"
)

future_forecast = pd.DataFrame({
    "Date": future_dates,
    "Forecast_Units": forecast_value
})

display(future_forecast.head(10))

future_forecast.to_csv("demand_forecast_30_days.csv", index=False)
print("Forecast saved successfully!")
```

### Prepare customer-level features

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score
import pandas as pd
import numpy as np

last_date = sales_clean["Order_Date"].max()

customer_features = sales_clean.groupby("Customer_ID").agg(
    Last_Purchase=("Order_Date", "max"),
    Orders=("Order_ID", "nunique"),
    Spend=("Sales", "sum"),
    Avg_Order_Value=("Sales", "mean")
).reset_index()

customer_features["Recency_Days"] = (
    last_date - customer_features["Last_Purchase"]
).dt.days

display(customer_features.head())
```

### Define a churn-risk label
For this demo, I will label customers who have not purchased during the last 90 days of the dataset as potentially churned. 

1 = potentially churned under our chosen definition.

0 = purchased within the last 90 days.

```python
print(customer_features.head())
print(customer_features.columns)
```

```python

customer_features["Churn_Risk_Label"] = (customer_features["Recency_Days"] > 90).astype(int)

print(customer_features["Churn_Risk_Label"].value_counts())

display(customer_features[["Customer_ID", "Recency_Days", "Churn_Risk_Label"]].head(10))
```

### Train the Random Forest model

```python

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    roc_auc_score
)

# Select features that do not directly define the churn label
features = ["Orders", "Spend", "Avg_Order_Value"]

X = customer_features[features]
y = customer_features["Churn_Risk_Label"]

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# Train the model
churn_model = RandomForestClassifier(
    n_estimators=150,
    random_state=42,
    class_weight="balanced"
)

churn_model.fit(X_train, y_train)

# Evaluate the model
predictions = churn_model.predict(X_test)
probabilities = churn_model.predict_proba(X_test)[:, 1]

print("Accuracy:", round(accuracy_score(y_test, predictions), 3))
print("\nClassification Report:")
print(classification_report(y_test, predictions, zero_division=0))
print("ROC-AUC:", round(roc_auc_score(y_test, probabilities), 3))
```

```python
customer_features.to_csv(
    "customer_churn_predictions.csv",
    index=False
)

print("Churn results saved successfully!")
```

### Inventory optimization
We'll recommend how many units each product should be reordered, 
based on historical daily demand, supplier lead time, and a seven-day safety buffer

### Calculate average daily demand

```python
product_demand = sales_clean.groupby("Product_ID").agg(
    Total_Units=("Quantity", "sum"),
    First_Order=("Order_Date", "min"),
    Last_Order=("Order_Date", "max")
).reset_index()

product_demand["Days_Observed"] = (
    product_demand["Last_Order"] - product_demand["First_Order"]
).dt.days + 1

product_demand["Avg_Daily_Demand"] = (
    product_demand["Total_Units"] /
    product_demand["Days_Observed"].clip(lower=1)
)

display(product_demand.head())
```

### Calculate recommended reorder quantities

```python
reorder = inventory.merge(
    product_demand[["Product_ID", "Avg_Daily_Demand"]],
    on="Product_ID",
    how="left"
)

reorder["Avg_Daily_Demand"] = (
    reorder["Avg_Daily_Demand"].fillna(0)
)

# Lead-time demand plus 7 days of safety stock
reorder["Target_Stock"] = np.ceil(
    reorder["Avg_Daily_Demand"] *
    (reorder["Lead_Time_Days"] + 7)
).astype(int)

reorder["Suggested_Reorder"] = (
    reorder["Target_Stock"] - reorder["Current_Stock"]
).clip(lower=0)

display(
    reorder.sort_values(
        "Suggested_Reorder",
        ascending=False
    ).head(15)
)
```

### Visualize the top 10 reorder recommendations

```python
# top_reorder = reorder.nlargest(10, "Suggested_Reorder")
top_reorder  = inventory.groupby("Product_ID")["Suggested_Reorder"].sum().reset_index().nlargest(10, "Suggested_Reorder")
top_10 = top_reorder.nlargest(10, "Suggested_Reorder")
plt.figure(figsize=(10, 5))
plt.bar(
    top_10["Product_ID"],
    top_10["Suggested_Reorder"]
)
plt.title("Top 10 Products — Suggested Reorder Quantity")
plt.xlabel("Product ID")
plt.ylabel("Units to Reorder")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
